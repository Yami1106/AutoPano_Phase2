```python
import cv2

img_path = "/home/rawbeat-ubuntux/vault/01_University/Semester_02/Computer-Vision/rbanerjee1_p1/Phase2/Data/Train/11.jpg"

# Read the image
img = cv2.imread(img_path)

# Check if image loaded correctly
if img is None:
    raise FileNotFoundError("Image not found or path is incorrect")

# Resize to 320x240 (width, height)
img_resized = cv2.resize(img, (320, 240))

# Optional: display the result
cv2.imshow("Resized Image", img_resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
